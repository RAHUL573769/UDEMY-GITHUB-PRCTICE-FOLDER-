```python
import numpy as np 
x=np.linspace(0,5,11)
y=x**2
```

```python
x,y
```

```python
import matplotlib.pyplot as plt 
plt.plot(x,y)

plt.xlabel("X-Label")
plt.ylabel("Y- Label")
plt.title("Title")
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'r')


plt.subplot(1,2,2)
plt.plot(y,x,'b')
```

```python
import matplotlib.pyplot as plt
figure = plt.figure()

axes=figure.add_axes([0.1,0,0.8,0.8])

axes.plot(x,y)

axes.set_xlabel("X Label")
axes.set_ylabel("Y Label")
axes.set_title("Set-Title")
```

```python
import matplotlib.pyplot as plt
figure = plt.figure()

axes=figure.add_axes([0.1,0,0.8,0.8])

fiq=plt.figure()

axes1=figure.add_axes([0.1,0,0.8,0.8])
axes1.plot(x,y)


axes1.set_title("1st plot")
axesy=figure.add_axes([0.6,0.9,0.8,0.4])
axesy.plot(y,x)
```

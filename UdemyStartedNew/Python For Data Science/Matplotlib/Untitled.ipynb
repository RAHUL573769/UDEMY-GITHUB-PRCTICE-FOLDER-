```python
import matplotlib.pyplot as plt
import numpy as np

# Create the first subplot in a 2x2 grid, at position 1 (top-left)
plt.subplot(2, 2, 1)
plt.plot(np.random.rand(10))
plt.title('Subplot 1')

# Create the second subplot in the same 2x2 grid, at position 2 (top-right)
plt.subplot(2, 2, 2)
plt.plot(np.sin(np.linspace(0, 2*np.pi, 100)))
plt.title('Subplot 2')

# Create the third subplot in the same 2x2 grid, at position 3 (bottom-left)
plt.subplot(2, 2, 3)
plt.bar(['A', 'B', 'C'], [10, 5, 15])
plt.title('Subplot 3')

# Create the fourth subplot in the same 2x2 grid, at position 4 (bottom-right)
plt.subplot(2, 2, 4)
plt.scatter(np.random.rand(20), np.random.rand(20))
plt.title('Subplot 4')

plt.tight_layout() # Adjust subplot parameters for a tight layout
plt.show()
```
